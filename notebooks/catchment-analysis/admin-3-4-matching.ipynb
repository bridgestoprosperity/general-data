Task 2: Convert bridge sites dataframe to geodataframe that has custom Rwanda TM CRS copied from village boundaries geodataframe

```python
# Convert bridge sites dataframe to geodataframe that has custom Rwanda TM CRS copied from village boundaries geodataframe
def map_bridges(bridges, bridges_lat, bridges_lon, village_boundaries):

    # Check CRS of village boundaries gdf
    print(f'The CRS of the village boundaries gdf is: {village_boundaries.crs}')

    # Create lat/lon variables
    lon = bridges[bridges_lon]
    lat = bridges[bridges_lat]

    # Create gdf of bridges data by converting lat/lon values to list of Shapely Point objects
    bridge_points = gpd.GeoDataFrame(bridges, geometry=gpd.points_from_xy(x=lon, y=lat), crs='EPSG:4326')

    # Set CRS of bridges gdf to CRS of village boundaries gdf
    bridge_points.to_crs(village_boundaries.crs, inplace=True)

    # Check that reprojection was successful
    print(f'The CRS of the bridges gdf is: {bridge_points.crs}')
    
    return bridge_points

bridge_points = map_bridges(bridges = Rwanda_bridges,
                            bridges_lat = "Lat",
                            bridges_lon = "Long",
                            village_boundaries = Rwanda_village_boundaries)
```

Task 3: Dissolve village boundaries gdf to create gdfs for admin level 3 and admin level 4 boundaries according to the following table:
| Admin level number | Admin level name | Column in village boundaries |
|-------------|--------------------------|-------------------------------|
| 5           | Village                  | Village_ID                    |
| 4           | Cell                     | Cell_ID                       |
| 3           | Sector                   | Sector_ID                     |
| 2           | District                 | Distr_ID                      |
| 1           | Province                 | Prov_ID                       |

```python
# Dissolve village boundaries gdf to create gdfs for admin 1-4 boundaries
def dissolve_boundaries(village_boundaries, admin_level):
    # Dissolve the village boundaries by the specified admin level
    dissolved_boundaries = village_boundaries.dissolve(by=admin_level)
    dissolved_boundaries.reset_index(inplace=True)

    # Remove unneeded admin columns (e.g., 'Village_ID' for admin level 4)
    if admin_level == 'Cell_ID':
        dissolved_boundaries.drop(columns=['Village_ID'], inplace=True)
    if admin_level == 'Sector_ID':
        dissolved_boundaries.drop(columns=['Village_ID', 'Cell_ID'], inplace=True)
    if admin_level == 'Dist_ID':
        dissolved_boundaries.drop(columns=['Village_ID', 'Cell_ID', 'Sector_ID'], inplace=True)
    elif admin_level == 'Prov_ID':
        dissolved_boundaries.drop(columns=['Village_ID', 'Cell_ID', 'Sector_ID', 'Distr_ID'], inplace=True)

    return dissolved_boundaries

admin_1_boundaries = dissolve_boundaries(village_boundaries=Rwanda_village_boundaries, admin_level='Prov_ID')
admin_2_boundaries = dissolve_boundaries(village_boundaries=Rwanda_village_boundaries, admin_level='Distr_ID')
admin_3_boundaries = dissolve_boundaries(village_boundaries=Rwanda_village_boundaries, admin_level='Sector_ID')
admin_4_boundaries = dissolve_boundaries(village_boundaries=Rwanda_village_boundaries, admin_level='Cell_ID')
```

Task 4: Calculate & assign centroids to polygons in all three gdfs (admin 3, 4 & 5)

```python
# Calculate the centroid of each polygon in the gdf
def calculate_admin_centroids(admin_gdf):
    admin_gdf['centroid'] = admin_gdf.geometry.centroid
    return admin_gdf

# Apply the function to calculate centroids
admin_5_centroids = calculate_admin_centroids(Rwanda_village_boundaries)
admin_4_centroids = calculate_admin_centroids(admin_4_boundaries)
admin_3_centroids = calculate_admin_centroids(admin_3_boundaries)
```

Task 5: Assign centroids to each catchment survey response by matching the admin codes from the catchment survey data to the admin codes in the appropriate admin gdf

```python
def assign_home_centroids(survey_df, admin_5_centroids, admin_4_centroids, admin_3_centroids):
    # Ensure admin codes are strings
    survey_df = survey_df.copy()
    survey_df['Home Village - Admin Code'] = survey_df['Home Village - Admin Code'].astype(str)
    admin_5_centroids = admin_5_centroids.copy()
    admin_4_centroids = admin_4_centroids.copy()
    admin_3_centroids = admin_3_centroids.copy()
    admin_5_centroids['Village_ID'] = admin_5_centroids['Village_ID'].astype(str)
    admin_4_centroids['Cell_ID'] = admin_4_centroids['Cell_ID'].astype(str)
    admin_3_centroids['Sector_ID'] = admin_3_centroids['Sector_ID'].astype(str)

    # Merge admin 5 (village) centroids
    df = survey_df.merge(
        admin_5_centroids[['Village_ID', 'centroid']],
        left_on='Home Village - Admin Code',
        right_on='Village_ID',
        how='left'
    ).rename(columns={'centroid': 'home_centroid_admin5'})

    # Merge admin 4 (cell) centroids
    df = df.merge(
        admin_4_centroids[['Cell_ID', 'centroid']],
        left_on='Home Village - Admin Code',
        right_on='Cell_ID',
        how='left'
    ).rename(columns={'centroid': 'home_centroid_admin4'})

    # Merge admin 3 (sector) centroids
    df = df.merge(
        admin_3_centroids[['Sector_ID', 'centroid']],
        left_on='Home Village - Admin Code',
        right_on='Sector_ID',
        how='left'
    ).rename(columns={'centroid': 'home_centroid_admin3'})

    # Combine into a single column, prioritizing admin5 > admin4 > admin3
    df['home_centroid'] = df['home_centroid_admin5'].combine_first(df['home_centroid_admin4']).combine_first(df['home_centroid_admin3'])

    # Drop intermediate columns
    df = df.drop(columns=['home_centroid_admin5', 'home_centroid_admin4', 'home_centroid_admin3', 'Village_ID', 'Cell_ID', 'Sector_ID'])

    return df

# Apply the function
Rwanda_catchment_surveys_with_home_centroids = assign_home_centroids(
    survey_df=Rwanda_catchment_surveys,
    admin_5_centroids=admin_5_centroids,
    admin_4_centroids=admin_4_centroids,
    admin_3_centroids=admin_3_centroids
)
```

```python
def assign_destination_centroids(survey_df, admin_5_centroids, admin_4_centroids, admin_3_centroids):
    # Ensure admin codes are strings
    survey_df = survey_df.copy()
    survey_df['Destination - Admin Code'] = survey_df['Destination - Admin Code'].astype(str)
    admin_5_centroids = admin_5_centroids.copy()
    admin_4_centroids = admin_4_centroids.copy()
    admin_3_centroids = admin_3_centroids.copy()
    admin_5_centroids['Village_ID'] = admin_5_centroids['Village_ID'].astype(str)
    admin_4_centroids['Cell_ID'] = admin_4_centroids['Cell_ID'].astype(str)
    admin_3_centroids['Sector_ID'] = admin_3_centroids['Sector_ID'].astype(str)

    # Merge admin 5 (village) centroids
    df = survey_df.merge(
        admin_5_centroids[['Village_ID', 'centroid']],
        left_on='Destination - Admin Code',
        right_on='Village_ID',
        how='left'
    ).rename(columns={'centroid': 'destination_centroid_admin5'})

    # Merge admin 4 (cell) centroids
    df = df.merge(
        admin_4_centroids[['Cell_ID', 'centroid']],
        left_on='Destination - Admin Code',
        right_on='Cell_ID',
        how='left'
    ).rename(columns={'centroid': 'destination_centroid_admin4'})

    # Merge admin 3 (sector) centroids
    df = df.merge(
        admin_3_centroids[['Sector_ID', 'centroid']],
        left_on='Destination - Admin Code',
        right_on='Sector_ID',
        how='left'
    ).rename(columns={'centroid': 'destination_centroid_admin3'})

    # Combine into a single column, prioritizing admin5 > admin4 > admin3
    df['destination_centroid'] = df['destination_centroid_admin5'].combine_first(df['destination_centroid_admin4']).combine_first(df['destination_centroid_admin3'])

    # Drop intermediate columns
    df = df.drop(columns=['destination_centroid_admin5', 'destination_centroid_admin4', 'destination_centroid_admin3', 'Village_ID', 'Cell_ID', 'Sector_ID'])

    return df

# Apply the function
Rwanda_catchment_surveys_all_centroids = assign_destination_centroids(
    survey_df=Rwanda_catchment_surveys_with_home_centroids,
    admin_5_centroids=admin_5_centroids,
    admin_4_centroids=admin_4_centroids,
    admin_3_centroids=admin_3_centroids
)
```

Task 6: Assign bridge coordinates to catchment survey responses

```python
# Assign bridge coordinates to catchment survey data
def assign_bridge_coordinates(survey_data, bridge_points):

    # Ensure both columns are of the same type before merging
    survey_data['Bridge Site'] = survey_data['Bridge Site'].astype(str)
    bridge_points['Bridge Name'] = bridge_points['Bridge Name'].astype(str)

    # Join bridge coordinates to survey data using shared bridge names
    survey_data = survey_data.merge(bridge_points[['Bridge Name', 'geometry']],
                                    left_on='Bridge Site',
                                    right_on='Bridge Name',
                                    how='left')

    # Rename coordinates column
    survey_data.rename(columns={'geometry': 'bridge_coordinates'}, inplace=True)

    return survey_data

# Apply the function
Rwanda_catchment_surveys_final = assign_bridge_coordinates(survey_data = Rwanda_catchment_surveys_all_centroids, bridge_points = bridge_points)
```

Task 7: Calculate Euclidean distance between home/destination centroids and bridge sites for each catchment survey response. Calculate home and destination distances separately and then sum for total distance

```python
# Calculate journey distances for survey respondents
def calculate_journey_distances(survey_data):

    # Calculate distance from home admin centroid to bridge coordinates
    survey_data['home_to_bridge_distance'] = survey_data.apply(
        lambda row: row['home_centroid'].distance(row['bridge_coordinates'])
        if row['home_centroid'] is not None and row['bridge_coordinates'] is not None
        else np.nan, axis=1)

    # Calculate distance from bridge coordinates to destination admin centroid
    survey_data['bridge_to_destination_distance'] = survey_data.apply(
        lambda row: row['bridge_coordinates'].distance(row['destination_centroid'])
        if row['bridge_coordinates'] is not None and row['destination_centroid'] is not None
        else np.nan, axis=1)

    survey_data['total_distance'] = survey_data['home_to_bridge_distance'] + survey_data['bridge_to_destination_distance']

    return survey_data

# Apply the function
Rwanda_catchment_surveys_final = calculate_journey_distances(survey_data = Rwanda_catchment_surveys_final)
```

Task 8: Explore journey distance distribution and outliers

```python
def calculate_summary_stats(survey_data):
    stats = survey_data.groupby('Bridge Site')['total_distance'].agg(['mean', 'median', 'std', 'count'])
    null_counts = survey_data.groupby('Bridge Site')['total_distance'].apply(lambda x: x.isnull().sum())
    stats['null_count'] = null_counts
    stats['total_count'] = survey_data.groupby('Bridge Site')['total_distance'].count() + null_counts
    stats = stats.reset_index()
    return stats

# Apply the function
average_distances = calculate_summary_stats(survey_data=Rwanda_catchment_surveys_final)

print(average_distances)
```

```python
import matplotlib.cm as cm

def plot_ridgeline_histograms(survey_data, bridge_col, distance_col, n_bins=100):
    """
    Plots ridgeline histograms of journey distances by bridge site.
    Returns: ordered_bridges, bridge_colors
    """
    # Copy data and convert distance to km
    data = survey_data[[bridge_col, distance_col]].dropna().copy()
    data["distance_km"] = data[distance_col] / 1000

    # Order bridges by median distance
    bridge_medians_km = data.groupby(bridge_col)["distance_km"].median()
    bridge_medians_sorted = bridge_medians_km.sort_values()
    ordered_bridges = bridge_medians_sorted.index.tolist()

    # Global bins
    global_min = data["distance_km"].min()
    global_max = data["distance_km"].max()
    bin_edges = np.linspace(global_min, global_max, n_bins + 1)

    fig, ax = plt.subplots(figsize=(10, 8))
    spacing = 1.0

    # Assign colors
    cmap = cm.get_cmap("tab20", len(ordered_bridges))
    bridge_colors = {bridge: cmap(i) for i, bridge in enumerate(ordered_bridges)}

    for i, bridge in enumerate(ordered_bridges):
        subset = data[data[bridge_col] == bridge]["distance_km"].values
        counts, _ = np.histogram(subset, bins=bin_edges, density=True)

        # Normalize height
        if counts.max() > 0:
            counts = counts / counts.max() * 0.9

        y_offset = i * spacing
        color = bridge_colors[bridge]

        # Draw ridgeline
        ax.fill_between(
            bin_edges[:-1],
            y_offset,
            counts + y_offset,
            step="post",
            color=color,
            alpha=0.85,
            edgecolor=None
        )

        # Median line
        median = bridge_medians_km[bridge]
        ax.plot([median, median], [y_offset, y_offset + 0.9],
                color="black", linewidth=1.5, linestyle="--")

    # Y-axis labels
    ax.set_yticks([i * spacing for i in range(len(ordered_bridges))])
    ax.set_yticklabels(ordered_bridges)
    ax.set_xlabel("Journey distance (km)")
    ax.set_title("Distribution of Journey Distances by Bridge Site")

    ax.grid(axis="x", alpha=0.3)
    ax.set_ylim(-spacing, spacing * len(ordered_bridges))

    plt.tight_layout()
    plt.show()

    return ordered_bridges, bridge_colors
```

```python
def plot_violin(survey_data, bridge_col, distance_col):
    """
    Plots violin plots of journey distances by bridge site,
    showing individual data points and the distribution shape.
    """
    # Copy data and convert distance to km
    data = survey_data[[bridge_col, distance_col]].dropna().copy()
    data["distance_km"] = data[distance_col] / 1000

    # Order bridges by median distance
    bridge_medians = data.groupby(bridge_col)["distance_km"].median()
    ordered_bridges = bridge_medians.sort_values(ascending=False).index.tolist()

    # Set up the figure
    plt.figure(figsize=(10, 8))
    
    # Violin plot with individual points
    sns.violinplot(
        x="distance_km",
        y=bridge_col,
        data=data,
        order=ordered_bridges,
        inner=None,  # No points inside violin, we'll add them separately
        palette="tab20",
        cut=0  # limits violins to min/max of data
    )

    # Add individual points with a stripplot
    sns.stripplot(
        x="distance_km",
        y=bridge_col,
        data=data,
        order=ordered_bridges,
        color="k",
        alpha=0.5,
        size=3,
        jitter=True
    )

    # Add median lines
    for i, bridge in enumerate(ordered_bridges):
        median_val = bridge_medians[bridge]
        plt.plot([median_val, median_val], [i - 0.4, i + 0.4],
                 color="white", linewidth=2, linestyle="--")

    plt.xlabel("Journey distance (km)")
    plt.ylabel("Bridge site")
    plt.title("Distribution of Journey Distances by Bridge Site")
    plt.grid(axis="x", alpha=0.3)
    plt.tight_layout()
    plt.show()

    return ordered_bridges
```

```python
def plot_violin_with_outlier_filter_dual(survey_data, bridge_col, distance_col, show_outliers=True):
    """
    Produces two violin plots:
    1. Original data with outliers filtered but optionally marked
    2. Clean violin plot of only inlier data (outliers removed)
    
    Features:
    - Outlier filtering per bridge (IQR method)
    - Halo effect for individual points
    - Median lines
    - Bridges ordered by median (largest median on top)
    """
    # Copy data and convert distance to km
    data = survey_data[[bridge_col, distance_col]].dropna().copy()
    data["distance_km"] = data[distance_col] / 1000

    # Order bridges by median distance (largest median on top)
    bridge_medians = data.groupby(bridge_col)["distance_km"].median()
    ordered_bridges = bridge_medians.sort_values(ascending=False).index.tolist()

    # Function to filter outliers per bridge
    def filter_outliers_iqr(subset):
        Q1 = subset.quantile(0.25)
        Q3 = subset.quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        inliers = subset[(subset >= lower) & (subset <= upper)]
        outliers = subset[(subset < lower) | (subset > upper)]
        return inliers, outliers

    # Prepare filtered data and outliers
    filtered_data = pd.DataFrame(columns=data.columns)
    outlier_data = pd.DataFrame(columns=data.columns)
    for bridge in ordered_bridges:
        subset = data[data[bridge_col] == bridge]["distance_km"]
        inliers, outliers = filter_outliers_iqr(subset)
        filtered_data = pd.concat([filtered_data, pd.DataFrame({bridge_col: bridge, "distance_km": inliers})])
        outlier_data = pd.concat([outlier_data, pd.DataFrame({bridge_col: bridge, "distance_km": outliers})])

    # ----- Figure 1: Violin plot with outliers marked -----
    plt.figure(figsize=(10, 8))
    sns.violinplot(
        x="distance_km",
        y=bridge_col,
        data=filtered_data,
        order=ordered_bridges,
        inner=None,
        palette="tab20",
        cut=0
    )

    # Add individual points with halo
    for i, bridge in enumerate(ordered_bridges):
        subset = filtered_data[filtered_data[bridge_col] == bridge]["distance_km"]
        y_positions = np.full_like(subset, i)
        y_positions += np.random.uniform(-0.1, 0.1, size=len(subset))
        # plt.scatter(subset, y_positions, color='white', s=20, edgecolors='none', alpha=0.5, zorder=1)
        plt.scatter(subset, y_positions, color='black', alpha=0.5, s=8, zorder=2)

    # Mark outliers
    if show_outliers and not outlier_data.empty:
        for i, bridge in enumerate(ordered_bridges):
            subset = outlier_data[outlier_data[bridge_col] == bridge]["distance_km"]
            if len(subset) > 0:
                y_positions = np.full_like(subset, i)
                y_positions += np.random.uniform(-0.1, 0.1, size=len(subset))
                plt.scatter(subset, y_positions, color='red', s=8, alpha=0.3, zorder=0)

    # Median lines
    filtered_medians = filtered_data.groupby(bridge_col)["distance_km"].median()
    for i, bridge in enumerate(ordered_bridges):
        median_val = filtered_medians[bridge]
        plt.plot([median_val, median_val], [i - 0.4, i + 0.4],
                 color='black', linewidth=2.5, linestyle='-', zorder=3)

    plt.xlabel("Total journey distance of survey respondent (km)")
    plt.ylabel("Bridge site")
    plt.grid(axis='x', alpha=0.3)
    plt.tight_layout()
    plt.show()

    # ----- Figure 2: Clean violin plot (outliers removed) -----
    plt.figure(figsize=(10, 8))
    sns.violinplot(
        x="distance_km",
        y=bridge_col,
        data=filtered_data,
        order=ordered_bridges,
        inner=None,
        palette="tab20",
        cut=0
    )

    # Add individual points with halo
    for i, bridge in enumerate(ordered_bridges):
        subset = filtered_data[filtered_data[bridge_col] == bridge]["distance_km"]
        y_positions = np.full_like(subset, i)
        y_positions += np.random.uniform(-0.1, 0.1, size=len(subset))
        # plt.scatter(subset, y_positions, color='white', s=20, edgecolors='none', alpha=0.5, zorder=1)
        plt.scatter(subset, y_positions, color='black', s=8, alpha=0.5, zorder=2)

    # Median lines
    for i, bridge in enumerate(ordered_bridges):
        median_val = filtered_medians[bridge]
        plt.plot([median_val, median_val], [i - 0.4, i + 0.4],
                 color='black', linewidth=2.5, linestyle='-', zorder=3)

    plt.xlabel("Total journey distance of survey respondent (km)")
    plt.ylabel("Bridge site")
    plt.grid(axis='x', alpha=0.3)
    plt.tight_layout()
    plt.show()

    return ordered_bridges, filtered_data, outlier_data
```

```python
def plot_violin_with_outlier_filter_dual_vertical(
    survey_data, bridge_col, distance_col, show_outliers=True
):

    # Copy data and convert distance to km
    data = survey_data[[bridge_col, distance_col]].dropna().copy()
    data["distance_km"] = data[distance_col] / 1000

    # Order bridges by median distance (largest median on right)
    bridge_medians = data.groupby(bridge_col)["distance_km"].median()
    ordered_bridges = bridge_medians.sort_values(ascending=False).index.tolist()

    # IQR filtering
    def filter_outliers_iqr(subset):
        Q1 = subset.quantile(0.25)
        Q3 = subset.quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        inliers = subset[(subset >= lower) & (subset <= upper)]
        outliers = subset[(subset < lower) | (subset > upper)]
        return inliers, outliers

    filtered_data = []
    outlier_data = []

    for bridge in ordered_bridges:
        subset = data[data[bridge_col] == bridge]["distance_km"]
        inliers, outliers = filter_outliers_iqr(subset)

        filtered_data.append(pd.DataFrame({
            bridge_col: bridge,
            "distance_km": inliers
        }))

        outlier_data.append(pd.DataFrame({
            bridge_col: bridge,
            "distance_km": outliers
        }))

    filtered_data = pd.concat(filtered_data)
    outlier_data = pd.concat(outlier_data)

    # ====================================================
    # FIGURE 1 — With Outliers Marked
    # ====================================================
    plt.figure(figsize=(12, 8))

    sns.violinplot(
        x=bridge_col,
        y="distance_km",
        data=filtered_data,
        order=ordered_bridges,
        inner=None,
        palette="tab20",
        cut=0
    )

    # Halo-style points (vertical jitter on x)
    for i, bridge in enumerate(ordered_bridges):
        subset = filtered_data[filtered_data[bridge_col] == bridge]["distance_km"]

        x_positions = np.full(len(subset), i, dtype=float)
        x_positions += np.random.uniform(-0.1, 0.1, size=len(subset))

        plt.scatter(x_positions, subset, color='black', alpha=0.5, s=8, zorder=2)

    # Mark outliers
    if show_outliers and not outlier_data.empty:
        for i, bridge in enumerate(ordered_bridges):
            subset = outlier_data[outlier_data[bridge_col] == bridge]["distance_km"]
            if len(subset) > 0:
                x_positions = np.full(len(subset), i, dtype=float)
                x_positions += np.random.uniform(-0.1, 0.1, size=len(subset))
                plt.scatter(x_positions, subset, color='red', s=8, alpha=0.3, zorder=0)

    # Median lines (horizontal now)
    filtered_medians = filtered_data.groupby(bridge_col)["distance_km"].median()

    for i, bridge in enumerate(ordered_bridges):
        median_val = filtered_medians[bridge]
        plt.plot([i - 0.4, i + 0.4], [median_val, median_val],
                 color='black', linewidth=2.5, linestyle='-', zorder=3)

    plt.xticks(rotation=45)
    plt.xlabel("Bridge site")
    plt.ylabel("Total journey distance of survey respondent (km)")
    plt.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    plt.show()

    # ====================================================
    # FIGURE 2 — Clean Version
    # ====================================================
    plt.figure(figsize=(12, 8))

    sns.violinplot(
        x=bridge_col,
        y="distance_km",
        data=filtered_data,
        order=ordered_bridges,
        inner=None,
        palette="tab20",
        cut=0
    )

    for i, bridge in enumerate(ordered_bridges):
        subset = filtered_data[filtered_data[bridge_col] == bridge]["distance_km"]

        x_positions = np.full(len(subset), i, dtype=float)
        x_positions += np.random.uniform(-0.1, 0.1, size=len(subset))

        plt.scatter(x_positions, subset, color='black', alpha=0.5, s=8, zorder=2)

        median_val = filtered_medians[bridge]
        plt.plot([i - 0.4, i + 0.4], [median_val, median_val],
                 color='black', linewidth=2.5, linestyle='-', zorder=3)

    plt.xticks(rotation=45)
    plt.xlabel("Bridge site")
    plt.ylabel("Total journey distance of survey respondent (km)")
    plt.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    plt.show()

    return ordered_bridges, filtered_data, outlier_data
```

```python
def plot_journey_paths_weighted(
    survey_data, home_col, bridge_col, destination_col,
    bridge_points, admin_1_boundaries, admin_2_boundaries,
    ordered_bridges, bridge_colors,
    min_width=1.0, max_width=5.0
):
    """
    Plots journey paths from home → bridge → destination.
    Features:
      - Line width ∝ sqrt(# respondents per journey)
      - Colors match ridgeline plot
      - Latitude/longitude labels removed
      - Legend in upper-left, ordered to match ridgeline bottom-to-top
    """

    plt.figure(figsize=(12, 12))
    ax = plt.subplot(111)

    # --- Drop rows with missing data ---
    valid_data = survey_data.dropna(subset=[home_col, bridge_col, destination_col]).copy()

    # --- Convert home and destination to tuples safely ---
    def to_tuple_safe(x):
        if x is None:
            return None
        elif hasattr(x, 'x') and hasattr(x, 'y'):
            return (x.x, x.y)
        else:
            return tuple(x)

    valid_data['home_tuple'] = valid_data[home_col].apply(to_tuple_safe)
    valid_data['dest_tuple'] = valid_data[destination_col].apply(to_tuple_safe)
    valid_data = valid_data.dropna(subset=['home_tuple', 'dest_tuple'])

    # --- Count duplicate journeys (home → bridge → destination) ---
    journey_counts = (
        valid_data
        .groupby([bridge_col, 'home_tuple', 'dest_tuple'])
        .size()
        .reset_index(name='count')
    )

    counts = journey_counts['count'].values
    sqrt_counts = np.sqrt(counts)

    # --- Scale sqrt counts to line widths ---
    if sqrt_counts.max() == sqrt_counts.min():  # avoid divide by zero
        widths = np.full_like(sqrt_counts, (min_width + max_width)/2)
    else:
        widths = min_width + (sqrt_counts - sqrt_counts.min()) / (sqrt_counts.max() - sqrt_counts.min()) * (max_width - min_width)

    # --- Plot each journey ---
    for i, row in journey_counts.iterrows():
        bridge_name = row[bridge_col]
        home = row['home_tuple']
        dest = row['dest_tuple']

        bridge_geom = bridge_points.loc[bridge_points['Bridge Name'] == bridge_name, 'geometry'].values[0]

        path = LineString([home, (bridge_geom.x, bridge_geom.y), dest])
        color = bridge_colors.get(bridge_name, 'gray')
        ax.plot(*path.xy, color=color, alpha=0.5, linewidth=widths[i])

    # --- Plot bridge points (without labels) ---
    for bridge_name in ordered_bridges:
        color = bridge_colors[bridge_name]
        bridge_point = bridge_points[bridge_points['Bridge Name'] == bridge_name]
        if not bridge_point.empty:
            ax.scatter(
                bridge_point.geometry.x,
                bridge_point.geometry.y,
                color=color,
                edgecolor='black',
                s=80,
                zorder=5
            )

    # --- Plot admin boundaries ---
    admin_1_boundaries.plot(ax=ax, color='none', edgecolor='black', alpha=1.0)
    admin_2_boundaries.plot(ax=ax, color='none', edgecolor='black', alpha=0.5)

    # --- Remove latitude/longitude labels and ticks ---
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_xlabel('')
    ax.set_ylabel('')
    ax.grid(False)

    # --- Legend: reversed to match ridgeline bottom-to-top ---
    reversed_bridges = list(reversed(ordered_bridges))
    handles = [plt.Line2D([0], [0], color=bridge_colors[bridge], lw=2) for bridge in reversed_bridges]
    labels = reversed_bridges
    ax.legend(handles=handles, labels=labels, title="Bridge Site",
              loc='upper left', bbox_to_anchor=(0.01, 0.99))

    ax.set_title("Journey Paths (Home to Bridge to Destination)")
    plt.tight_layout()
    plt.show()
```

```python
# Investigate whether the difference in journey distance for men and women is statistically significant
# Use a t-test to compare the means of two independent samples

def t_test_gender(survey_data):
    men_distances = survey_data[survey_data['Gender'] == 'Male']['total_distance'].dropna()
    women_distances = survey_data[survey_data['Gender'] == 'Female']['total_distance'].dropna()

    t_stat, p_value = stats.ttest_ind(men_distances, women_distances, equal_var=False)

    print(f"T-test results for journey distance by gender:")
    print(f"t_stat: {t_stat}")
    print(f"p_value: {p_value}")
    if p_value < 0.05:
        print("The difference in journey distance between men and women is statistically significant.")
    else:
        print("The difference in journey distance between men and women is not statistically significant.")

    return t_stat, p_value

t_test_gender_result = t_test_gender(Rwanda_catchment_surveys_final)
```

```python
# Investigate whether the difference in journey distance for men and women is statistically significant
# Use a Mann-Whitney U test to compare the distributions of two independent samples (non-parametric)

def mann_whitney_gender(survey_data, distance_col='total_distance'):
    
    # Extract groups
    men = survey_data[survey_data['Gender'] == 'Male'][distance_col].dropna()
    women = survey_data[survey_data['Gender'] == 'Female'][distance_col].dropna()
    
    # Sample sizes
    n_men = len(men)
    n_women = len(women)
    
    # Medians
    median_men = np.median(men)
    median_women = np.median(women)
    
    # Mann–Whitney U test (two-sided)
    u_stat, p_value = stats.mannwhitneyu(men, women, alternative='two-sided')
    
    # Rank-biserial correlation (effect size)
    rbc = 1 - (2 * u_stat) / (n_men * n_women)
    
    # Print results
    print("Mann–Whitney U Test: Journey Distance by Gender")
    print("------------------------------------------------")
    print(f"Sample size (men): {n_men}")
    print(f"Sample size (women): {n_women}")
    print(f"Median (men): {median_men:.2f}")
    print(f"Median (women): {median_women:.2f}")
    print(f"U statistic: {u_stat:.2f}")
    print(f"p-value: {p_value:.4f}")
    print(f"Rank-biserial correlation (effect size): {rbc:.3f}")
    
    if p_value < 0.05:
        print("Result: The difference between groups is statistically significant.")
    else:
        print("Result: The difference between groups is not statistically significant.")
    
    return u_stat, p_value, rbc
```

```python
# Investigate whether the difference in journey distance for other response variables with more than two categories is statistically significant
# Use ANOVA (Analysis of Variance) for this

def anova_test(survey_data, column_name):
    if column_name not in survey_data.columns:
        raise ValueError(f"Column '{column_name}' does not exist in the survey data.")

    groups = [group['total_distance'].dropna() for name, group in survey_data.groupby(column_name)]
    f_stat, p_value = stats.f_oneway(*groups)

    print(f"ANOVA results for journey distance by {column_name}:")
    print(f"f_stat: {f_stat}")
    print(f"p_value: {p_value}")
    if p_value < 0.05:
        print(f"At least one category of {column_name} has a statistically significant difference in journey distance.")
    else:
        print(f"The difference in journey distance between categories of {column_name} is not statistically significant.")

    return f_stat, p_value
```

```python
# Perform pairwise t-tests between all groups in 'Purpose of Travel - Categorized' and apply Bonferroni correction
from itertools import combinations
from scipy.stats import ttest_ind
from statsmodels.stats.multitest import multipletests

# Get unique groups
groups = Rwanda_catchment_surveys_final['Purpose of Travel - Categorized'].dropna().unique()

# Prepare lists to store results
p_values = []
comparisons = []

# Perform pairwise t-tests
for g1, g2 in combinations(groups, 2):
    data1 = Rwanda_catchment_surveys_final[Rwanda_catchment_surveys_final['Purpose of Travel - Categorized'] == g1]['total_distance'].dropna()
    data2 = Rwanda_catchment_surveys_final[Rwanda_catchment_surveys_final['Purpose of Travel - Categorized'] == g2]['total_distance'].dropna()
    stat, p = ttest_ind(data1, data2, equal_var=False)
    p_values.append(p)
    comparisons.append(f"{g1} vs {g2}")

# Apply Bonferroni correction
reject, pvals_corrected, _, _ = multipletests(p_values, alpha=0.05, method='bonferroni')

# Print pairs for which the difference is statistically significant, including p-values
print("Statistically significant pairs (Bonferroni corrected):")
for comp, pval_corr in zip(comparisons, pvals_corrected):
    if pval_corr < 0.05:
        print(f"{comp}: Bonferroni corrected p-value={pval_corr:.4g}")
    else:
        print(f"{comp}: Not statistically significant (p-value={pval_corr:.4g})")
```

```python
from statsmodels.stats.multitest import multipletests

def kruskal_wallis_with_posthoc(
    df,
    group_col,
    value_col,
    alpha=0.05,
    p_adjust="holm",          # "holm" (good default), or "bonferroni", "fdr_bh"
    min_group_n=5,
    use_dunn_if_available=True
):
    """
    Kruskal–Wallis H-test + post-hoc pairwise comparisons.

    Parameters
    ----------
    df : pandas.DataFrame
    group_col : str   categorical grouping variable (e.g., Purpose, Age category)
    value_col : str   numeric outcome (e.g., total_distance or distance_km)
    alpha : float
    p_adjust : str    correction method for multiple comparisons (statsmodels multipletests)
    min_group_n : int exclude tiny groups to avoid unstable p-values
    use_dunn_if_available : bool
        If True and scikit-posthocs is installed, runs Dunn's test (recommended).
        Otherwise falls back to pairwise Mann–Whitney U tests with p-value correction.

    Returns
    -------
    results : dict with:
      - 'kruskal': dict of H, p, df
      - 'group_summary': DataFrame (n, median, IQR)
      - 'posthoc': DataFrame of pairwise results (or Dunn matrix if Dunn available)
      - 'method': str describing posthoc method used
    """
    data = df[[group_col, value_col]].dropna().copy()

    # Drop rare groups
    counts = data[group_col].value_counts()
    keep_groups = counts[counts >= min_group_n].index
    data = data[data[group_col].isin(keep_groups)]

    # Build groups list for Kruskal–Wallis
    groups = []
    labels = []
    for g, sub in data.groupby(group_col, sort=False):
        vals = sub[value_col].values
        if len(vals) >= min_group_n:
            groups.append(vals)
            labels.append(g)

    if len(groups) < 2:
        raise ValueError("Not enough groups with sufficient data to run Kruskal–Wallis.")

    # Kruskal–Wallis
    H, p = stats.kruskal(*groups)

    # Effect size (epsilon-squared), common for Kruskal–Wallis
    n_total = sum(len(g) for g in groups)
    k = len(groups)
    eps_sq = (H - k + 1) / (n_total - k) if (n_total - k) > 0 else np.nan

    # Group summary: n, median, IQR
    # Sort by median descending
    def iqr(x):
        return np.nanpercentile(x, 75) - np.nanpercentile(x, 25)

    group_summary = (
        data.groupby(group_col)[value_col]
        .agg(n="count", median="median", q1=lambda x: np.nanpercentile(x, 25),
             q3=lambda x: np.nanpercentile(x, 75), iqr=iqr)
        .reset_index()
        .rename(columns={group_col: "group"})
        .sort_values("median", ascending=False)
        .reset_index(drop=True)
    )

    # Convert to km and round numeric columns to two decimal places
    group_summary = group_summary.assign(
        median=lambda x: x["median"] / 1000,
        q1=lambda x: x["q1"] / 1000,
        q3=lambda x: x["q3"] / 1000,
        iqr=lambda x: x["iqr"] / 1000
    ).round(2)

    # -------------------------
    # Post-hoc comparisons
    # -------------------------
    posthoc_method = None
    posthoc = None

    # Try Dunn's test first (recommended for Kruskal)
    if use_dunn_if_available:
        try:
            import scikit_posthocs as sp

            # Dunn test with chosen p-adjust method
            # scikit-posthocs supports: 'bonferroni', 'holm', 'fdr_bh', etc.
            dunn = sp.posthoc_dunn(
                data,
                val_col=value_col,
                group_col=group_col,
                p_adjust=p_adjust
            )

            # Reorder matrix to match labels
            dunn = dunn.loc[labels, labels]
            posthoc = dunn
            posthoc_method = f"Dunn's test with {p_adjust} correction (scikit-posthocs)"

        except ImportError:
            posthoc_method = "Pairwise Mann–Whitney U with p-value correction (fallback; install scikit-posthocs for Dunn)"

    # Fallback: pairwise Mann–Whitney U + correction
    if posthoc is None:
        pairs = []
        raw_p = []

        for i in range(len(labels)):
            for j in range(i + 1, len(labels)):
                g1, g2 = labels[i], labels[j]
                x = data.loc[data[group_col] == g1, value_col].values
                y = data.loc[data[group_col] == g2, value_col].values

                # Two-sided Mann–Whitney
                U, p_ij = stats.mannwhitneyu(x, y, alternative="two-sided")
                pairs.append((g1, g2, U, len(x), len(y)))
                raw_p.append(p_ij)

        reject, p_adj, _, _ = multipletests(raw_p, alpha=alpha, method=p_adjust)

        posthoc = pd.DataFrame(
            pairs,
            columns=["group1", "group2", "U", "n1", "n2"]
        )
        posthoc["p_raw"] = raw_p
        posthoc["p_adj"] = p_adj
        posthoc["significant"] = reject

        # Add median difference for interpretability (group1 - group2)
        med = data.groupby(group_col)[value_col].median().to_dict()
        posthoc["median_diff"] = posthoc["group1"].map(med) - posthoc["group2"].map(med)

    results = {
        "kruskal": {
            "H": float(H),
            "p": float(p),
            "df": int(k - 1),
            "epsilon_sq": float(eps_sq)
        },
        "group_summary": group_summary,
        "posthoc": posthoc,
        "method": posthoc_method or "Dunn's test"
    }

    return results
```

```python
def calculate_mean_distance_by_purpose_and_bridge(survey_data):
    # Group by Bridge Site and Purpose of Travel - Categorized, then calculate mean
    mean_distance = (
        survey_data
        .groupby(['Bridge Site', 'Purpose of Travel - Categorized'])['total_distance']
        .mean()
        .reset_index()
        .sort_values(['Bridge Site', 'total_distance'], ascending=[True, False])
    )

    # Scatterplot: X = Purpose of Travel, Y = Mean Journey Distance, color = Bridge Site
    plt.figure(figsize=(14, 7))
    categories = mean_distance['Purpose of Travel - Categorized'].dropna().unique()
    category_map = {cat: idx for idx, cat in enumerate(categories)}
    bridges = mean_distance['Bridge Site'].unique()
    for bridge in bridges:
        subset = mean_distance[mean_distance['Bridge Site'] == bridge]
        x = subset['Purpose of Travel - Categorized'].map(category_map)
        plt.scatter(
            x,
            subset['total_distance'],
            label=bridge,
            alpha=0.8,
            s=60
        )
    plt.title("Mean Journey Distance by Purpose of Travel and Bridge Site")
    plt.xlabel("Purpose of Travel - Categorized")
    plt.ylabel("Mean Journey Distance (m)")
    plt.xticks(ticks=range(len(categories)), labels=categories, rotation=45)
    plt.legend(title="Bridge Site", bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.show()

    return mean_distance

# Apply the function
calculate_mean_distance_by_purpose_and_bridge(Rwanda_catchment_surveys_final)
```
